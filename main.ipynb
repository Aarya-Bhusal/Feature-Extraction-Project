# Feature Extraction Project

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('anime.csv')
df.head()
```

**Make a new column for episode count**

```python
def extract_eps(title):
    take_eps = False
    eps = ""
    for i in title:
        if i == ')':
            take_eps = False
        if take_eps:
            eps = eps+i
        if i == '(':
            take_eps = True
            
    return eps.replace('eps','').strip()
df["Episodes"] = df["Title"].apply(extract_eps)
df["Episodes"] = df["Episodes"].astype('Int64')
df.head()
```

**Make a new column for time stamp**

```python
def extract_time(title):
    time_stamp = ""
    for i in range(len(title)):
        if(title[i] == ")"):
            for j in range(1, 20):
                time_stamp = time_stamp + title[i+j]
            return time_stamp
df["Time Stamp"] = df["Title"].apply(extract_time)

def extract_months(time_stamp):
    start = pd.to_datetime(time_stamp.split(" - ")[0], format="%b %Y")
    end = pd.to_datetime(time_stamp.split(" - ")[1], format="%b %Y")
    total_months = (end.year - start.year) * 12 + (end.month - start.month)
    return total_months
df["Months"] = df["Time Stamp"].apply(extract_months)

df.head()
```

**Find which anime has the highest score**

```python
df[df["Score"]==df["Score"].max()]
```
